## EDA and Feature Engineering of the dataset

Aim: Learn while performing EDA and feature engineering on this dataset, to understand the processes before a data can be used for modeling or prediction

### FEATURES
The various features of the cleaned dataset are explained below:
1) Airline: The name of the airline company is stored in the airline column. It is a categorical feature having 6 different airlines.
2) Flight: Flight stores information regarding the plane's flight code. It is a categorical feature.
3) Source City: City from which the flight takes off. It is a categorical feature having 6 unique cities.
4) Departure Time: This is a derived categorical feature obtained created by grouping time periods into bins. It stores information about the departure time and have 6 unique time labels.
5) Stops: A categorical feature with 3 distinct values that stores the number of stops between the source and destination cities.
6) Arrival Time: This is a derived categorical feature created by grouping time intervals into bins. It has six distinct time labels and keeps information about the arrival time.
7) Destination City: City where the flight will land. It is a categorical feature having 6 unique cities.
8) Class: A categorical feature that contains information on seat class; it has two distinct values: Business and Economy.
9) Duration: A continuous feature that displays the overall amount of time it takes to travel between cities in hours.
10)Days Left: This is a derived characteristic that is calculated by subtracting the trip date by the booking date.
11) Price: Target variable stores information of the ticket price.

```python
# All the libraries imported will be here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import openpyxl
from sklearn.preprocessing import OneHotEncoder
%matplotlib inline 
# matplotlib inline is a magic command that configures Matplotlib to display plots directly beneath your code cells inside Jupyter Notebooks or IPython interfaces, instead of opening them in a separate pop-up window
```

```python
df=pd.read_excel("flight_price.xlsx")
df.head()
```

```python
""" here we can see many features, each column has its own varied unclean data which we will be dealing with column by column"""
```

```python
#getting basic info of the data
df.info()
```

```python
# we can see there are 11 columns with 10683 entries and all are object type only last one is integer type.
df.head()
# now we will take each feature one by one
```

```python
# starting Feature Engineering with date of journey
df['Date']=df['Date_of_Journey'].str.split("/").str[0] #storing the DD in the date column
df['Month']=df['Date_of_Journey'].str.split("/").str[1] #storing the MM in the month column
df['Year']=df['Date_of_Journey'].str.split("/").str[2] #storing the YYYY in the date column
df.head()
```

```python
df.info()
# we can observe that the tree new columns created are also object type, we will be converting them to int class
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
df.info()
```

```python
# now we can drom the column "Date_of_Journey"
df.drop('Date_of_Journey', axis=1, inplace=True)
df.head()
```

```python
# now handling the "Arrival_Time"
'''
df['Arrival_hour']=df['Arrival_Time'].str.split(":").str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(":").str[1]
df.head(2)
'''
# applying this as is would create a problem wherein in Arrival_min we will also be getting some extra data to deal with that we will be devising a lambda function
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0]) #deals with extra samaan 
df['Arrival_hour']=df['Arrival_Time'].str.split(":").str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(":").str[1]
df.head(2)
```

```python
df["Arrival_hour"]=df['Arrival_hour'].astype(int)
df["Arrival_min"]=df['Arrival_min'].astype(int)
```

```python
df.drop('Arrival_Time',axis=1, inplace=True)
df.head(2)
```

```python
# similarly with departure time
df['Dep_Time']=df['Dep_Time'].apply(lambda x:x.split(' ')[0]) #deals with extra samaan 
df['Dep_hour']=df['Dep_Time'].str.split(":").str[0]
df['Dep_min']=df['Dep_Time'].str.split(":").str[1]
df["Dep_hour"]=df['Dep_hour'].astype(int)
df["Dep_min"]=df['Dep_min'].astype(int)
df.head(2)
```

```python
df.drop('Dep_Time',axis=1, inplace=True)
df.head(2)
```

```python
df.info()
```

```python
# Let us now deal with some categorical features, starting with Total_Stops
df.Total_Stops.unique()
```

```python
# to check how many null values are there
df[df['Total_Stops'].isnull()]
```

```python
# we can see only one Nan value is there and this variable we can replace with the mode value (generaly how categoricals are dealt)
df['Total_Stops'].mode()
```

```python
# dealing with all variables including nan we used this system here as from general knowledge we know more the number of stops higher might be the price
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3,'4 stops':4, np.nan:1}) 
```

```python
df[df['Total_Stops'].isnull()]
```

```python
# we can drop route as we have source and destination and also the time of journey so redundant
df.drop("Route", axis=1, inplace=True)
```

```python
df.head(2)
```

```python
# handling duration

#in our duration column we have one pplace where the duration in only 5m thus our following codes might not work:
'''
df['duration_hr']=df['Duration'].str.split(' ').str[0].str.split('h').str[0].fillna('0')
df['duration_min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0].fillna('0')
df['duration_hr']=df['duration_hr'].astype(int)
df['duration_min']=df['duration_min'].astype(int)
df.head(2)
'''

df['Duration']=np.where(df['Duration']=='5m','0h 5m',df['Duration'])
df[df['Duration']=='5m']
```

```python
# As we can see we have handled the isuue now we can use our original code
df['duration_hr']=df['Duration'].str.split(' ').str[0].str.split('h').str[0].fillna('0')
df['duration_min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0].fillna('0')
df['duration_hr']=df['duration_hr'].astype(int)
df['duration_min']=df['duration_min'].astype(int)
df.head(2)
```

```python
df['Total_Duration']=(df['duration_hr']*60)+df['duration_min']
#this will give us the total mins taken by our flight which is a much better indicator
df.head(2)
```

```python
df.info()
```

```python
#nor let us get rid of redundant colums
df.drop(['Duration','duration_hr','duration_min'], axis=1, inplace=True)
df.info()
```

```python
# remainingn variables and their values which are mostly categorical only

print(df.Airline.unique())
print(df.Source.unique())
print(df.Destination.unique())
print(df.Additional_Info.unique())
```

```python
#since they are not very large data we will be using oneHotEncoder for them
encoder=OneHotEncoder()
```

```python
encoder.fit_transform(df[['Airline','Source','Destination']]).toarray()
```

```python
df['Additional_Info']=np.where(df['Additional_Info']=='No info','No Info',df['Additional_Info'])
encoded_df=pd.DataFrame(encoder.fit_transform(df[['Airline','Source','Destination','Additional_Info']]).toarray(),columns=encoder.get_feature_names_out())
encoded_df.info()
```

```python
#merging the encoded data to our original dataframe
df=pd.merge(df,encoded_df,left_index=True,right_index=True)
df.head()
```

```python
#Finally getting rid of all the categorical data
df.drop(['Airline','Source','Destination','Additional_Info'],axis=1,inplace=True)
```
